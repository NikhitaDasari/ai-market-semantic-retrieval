```python
import requests

url = "https://api.weather.gov/points/41.8781,-87.6298"

headers = {
    "User-Agent": "weather-intelligence-homework"
}

response = requests.get(url, headers=headers)

print(response.status_code)
print(response.json())
```

```python
data = response.json()

print(data["properties"]["gridId"])
print(data["properties"]["gridX"])
print(data["properties"]["gridY"])
print(data["properties"]["forecast"])
print(data["properties"]["forecastHourly"])
```

```python
from weather_client import WeatherClient

client = WeatherClient()

latitude = 41.8781
longitude = -87.6298

point = client.get_point(latitude, longitude)

print(point["properties"]["gridId"])
print(point["properties"]["gridX"])
print(point["properties"]["gridY"])
print(point["properties"]["forecast"])
```

```python
forecast = client.get_forecast(latitude, longitude)

periods = forecast["properties"]["periods"]

for period in periods[:3]:
    print(period["name"])
    print(period["detailedForecast"])
    print()
```

```python
alerts = client.get_alerts(latitude, longitude)

print("Alert count:", len(alerts))

for alert in alerts[:3]:
    print(alert["properties"]["event"])
    print(alert["properties"]["description"])
    print()
```
